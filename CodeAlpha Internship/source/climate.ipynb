```python
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
all_disasters = [] 

# Scraping the first 3 pages as a test
for page_num in range(1, 150):
    url = f"https://reliefweb.int/disasters?page={page_num}"
    print(f"Scraping page {page_num}...")
    
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        print(f'{page_num} is working')
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 1. titles are stored inside <h3> tags
        # We locate all h3 tags on the page
        title_tags = soup.find_all('h3') 
        
        for tag in title_tags:
            # Get the raw text of the title (e.g., "Colombia: Earthquake - Aug 2024")
            raw_title = tag.text.strip()
            
            # Skip any h3 tags that aren't actually disaster titles 
            if ":" not in raw_title or "-" not in raw_title:
                continue 
                
            try:
                # 2.we Split the string into usable data!
                # First, split by ":" to separate the Country from the rest of the data's
                country = raw_title.split(':')[0].strip()
                
                # Next, we take the second section and split by "-" to separate Disaster from Date
                remainder = raw_title.split(':')[1]
                disaster_type = remainder.split('-')[0].strip()
                date = remainder.split('-')[1].strip()
                
                # 3. Add these data's to our all_disasters list
                all_disasters.append([country, disaster_type, date, raw_title])
                
            except IndexError:
                # If a title is formatted incorrectly, we continue 
                continue
                
    time.sleep(2) 

print("Scraping complete!")
```

```python
df = pd.DataFrame(all_disasters, columns=['Country', 'Disaster_Type', 'Date', 'Original_Title'])

df.head(10)
```
